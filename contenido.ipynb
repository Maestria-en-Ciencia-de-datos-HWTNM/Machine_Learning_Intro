## 4. Introducción a Machine Learning

### 4.1. Inteligencia Artificial vs Machine Learning vs Deep Learning

- **Inteligencia Artificial (IA)**: Conjunto de tecnologías que permiten a las máquinas realizar tareas que requieren inteligencia humana.
- **Machine Learning (ML)**: Subcampo de la IA que estudia algoritmos que aprenden y se adaptan a partir de datos.
- **Deep Learning (DL)**: Subcampo del ML que utiliza redes neuronales profundas.

### 4.2. ¿Qué es Machine Learning?

> *"Machine Learning es un método de análisis de datos que automatiza la creación de modelos analíticos. Es una rama de la IA basada en la idea de que los sistemas pueden aprender de los datos, identificar patrones y tomar decisiones con mínima intervención humana."* — SAS

> *"Machine Learning es la ciencia que permite que las computadoras actúen sin ser programadas explícitamente."* — Andrew Ng

### 4.3. Tipos de Datos

**Datos no orientados a dependencias** (estructura tabular):
- Columnas: atributos, características, variables.
- Filas: instancias, observaciones, ejemplos.


## 5. Python para Machine Learning

Python es el lenguaje más utilizado en ML por varias razones:
- Es fácil de aprender.
- Tiene un ecosistema extenso de bibliotecas.
- Es multiplataforma.
- Permite construir pipelines completos listos para producción.

### 5.1. Jupyter Notebooks

Jupyter Notebooks son documentos interactivos que combinan:
- Código ejecutable.
- Texto formateado (Markdown).
- Visualizaciones.

**Ventajas:**
- Ejecución directa en el navegador.
- Resultados inmediatos.
- Reproducibilidad.
- Exportación a `.py`.

## 6. Entorno de Trabajo

### 6.1. Anaconda

### 6.2. Google Colab
Alternativa en la nube: https://colab.research.google.com/

### 6.3. Blinder
Herramienta que convierte un repositorio de código (como uno de GitHub) en un entorno computacional interactivo.

### 6.4. Scikit-learn
Biblioteca principal para ML en Python. Incluye algoritmos para:
- Clasificación
- Regresión
- Clustering
- Reducción de dimensionalidad
- Preprocesamiento
- Selección de modelos
- Evaluación

### 6.5. PyCaret
PyCaret es una biblioteca de Machine Learning de bajo código que automatiza el flujo de trabajo de ciencia de datos:
- Clasificación
- Regresión
- Clustering
